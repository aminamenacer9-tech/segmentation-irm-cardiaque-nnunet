```python
import os
import pandas as pd

# === Dossier contenant toutes les images ===
image_dir = "/home/amenacer/Stage/Data/nnUnet/nnUNet_raw/Dataset010_MICE_HEART_2D/imagesTr/"

# === Fonction pour déterminer groupe et source ===
def classify_image(filename):
    if "CTRL_" in filename or "HFS_" in filename:
        return 1, "Rat", ""
    elif "MICE-HEART_" in filename:
        return 2, "CardiaTeam", ""
    elif "TollC" in filename:
        if "_z02" in filename:
            return 3, "Tollsome", "z02"
        elif "_z06" in filename:
            return 3, "Tollsome", "z06"
        else:
            return 3, "Tollsome", ""
    else:
        return "?", "Inconnu", ""

# === Création de la table ===
data = []
for f in os.listdir(image_dir):
    if f.endswith(".nii") or f.endswith(".nii.gz"):
        groupe, source, z = classify_image(f)
        data.append({
            "Nom de l’image": f,
            "Groupe": groupe,
            "z": z,
            "Source": source,
            "Utiliser pour retrain ?": "",
            "Commentaire": ""
        })

# === Export Excel ===
df = pd.DataFrame(data)
df.sort_values(by=["Groupe", "Nom de l’image"], inplace=True)  # tri optionnel
output_path = os.path.join(image_dir, "tableau_images_entraînement.xlsx")
df.to_excel(output_path, index=False)

print(f"✅ Fichier Excel créé ici : {output_path}")
```
