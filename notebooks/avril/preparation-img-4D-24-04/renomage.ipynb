```python
import os
import shutil

# Dossiers sources
raw_images = "/home/amenacer/Stage/Data/nnUnet/nnUNet_raw/imagesTr"
raw_labels = "/home/amenacer/Stage/Data/nnUnet/nnUNet_raw/labelsTr"

# Dossiers de destination nnU-Net
dest_images = "/home/amenacer/Stage/Data/nnUnet/nnUNet_raw/Dataset010_MICE_HEART_2D/imagesTr"
dest_labels = "/home/amenacer/Stage/Data/nnUnet/nnUNet_raw/Dataset010_MICE_HEART_2D/labelsTr"

# Création des dossiers si besoin
os.makedirs(dest_images, exist_ok=True)
os.makedirs(dest_labels, exist_ok=True)

def format_filename(fname, is_image=True):
    if ("3DplusT" not in fname) or (("_Original" not in fname) and ("_FUSION" not in fname)):
        return None  # ignorer les fichiers non ciblés
    name = fname.replace("_Original", "").replace("_FUSION", "")
    name = name.replace("3DplusT", "3D").replace(".nii.gz", "")
    return f"{name}_0000.nii.gz" if is_image else f"{name}.nii.gz"

# Images
for fname in os.listdir(raw_images):
    if fname.endswith(".nii.gz"):
        new_name = format_filename(fname, is_image=True)
        if new_name:
            shutil.copy(os.path.join(raw_images, fname), os.path.join(dest_images, new_name))
            print(f"[Image OK] {fname} -> {new_name}")
        else:
            print(f"[Image Ignored] {fname}")

# Masques
for fname in os.listdir(raw_labels):
    if fname.endswith(".nii.gz"):
        new_name = format_filename(fname, is_image=False)
        if new_name:
            shutil.copy(os.path.join(raw_labels, fname), os.path.join(dest_labels, new_name))
            print(f"[Label OK] {fname} -> {new_name}")
        else:
            print(f"[Label Ignored] {fname}")

print("\n✅ Script terminé.")
```

```python
import os
import re

label_dir = "/home/amenacer/Stage/Data/nnUnet/nnUNet_raw/Dataset010_MICE_HEART_2D/labelsTr"

for fname in os.listdir(label_dir):
    if fname.endswith(".nii.gz") and "(" in fname:
        # Supprimer la partie entre parenthèses (copie 1), (copier 2), etc.
        clean_name = re.sub(r"\s*\(.*?\)", "", fname)
        src = os.path.join(label_dir, fname)
        dst = os.path.join(label_dir, clean_name)
        os.rename(src, dst)
        print(f"[Renommé] {fname} → {clean_name}")

print("✅ Tous les noms 'copie' nettoyés.")
```
