```python
import os
import nibabel as nib

# === PARAMÈTRES DU GROUPE G1 ===
input_dir = "/home/amenacer/Stage/data-4D/hadia/TollSome nifti/G3"
output_dir = "/home/amenacer/Stage/data-4D/hadia/TollSome nifti/G3_3D"
os.makedirs(output_dir, exist_ok=True)

# === TRAITEMENT DES FICHIERS 4D ===
nifti_files = [f for f in os.listdir(input_dir) if f.endswith(".nii") or f.endswith(".nii.gz")]
print(f"🧠 {len(nifti_files)} fichier(s) à traiter dans {input_dir}")

for filename in nifti_files:
    filepath = os.path.join(input_dir, filename)
    print(f"📂 Lecture : {filename}")

    img = nib.load(filepath)
    data = img.get_fdata()
    affine = img.affine

    if data.ndim != 4:
        print(f"⛔ Fichier ignoré (non 4D) : {filename}")
        continue

    x, y, z, t = data.shape
    print(f"→ Dimensions : ({x}, {y}, {z}, {t})")

    for z_idx in range(z):
        slice_3dT = data[:, :, z_idx, :]  # shape (x, y, T)
        new_img = nib.Nifti1Image(slice_3dT, affine)

        base_name = os.path.splitext(os.path.splitext(filename)[0])[0]
        new_filename = f"{base_name}_z{z_idx:02d}_3DplusT.nii.gz"
        output_path = os.path.join(output_dir, new_filename)

        nib.save(new_img, output_path)
        print(f"✅ Sauvegardé : {new_filename}")

print("🎉 Terminé pour G3 !")
```

```python
import os
import nibabel as nib
import numpy as np
import shutil

# === Dossiers à modifier ===
input_dir = "/home/amenacer/Stage/data-4D/hadia/nifti-mask/G3_decoupe_squeez"
output_dir = "/home/amenacer/Stage/data-4D/hadia/nifti-mask/G3_non_vides"
os.makedirs(output_dir, exist_ok=True)

# === Filtrage des masques ===
nb_conservés = 0
nb_vides = 0

for filename in os.listdir(input_dir):
    if filename.endswith(".nii") or filename.endswith(".nii.gz"):
        filepath = os.path.join(input_dir, filename)
        img = nib.load(filepath)
        data = img.get_fdata()

        # Vérifier si le masque est complètement nul
        if np.count_nonzero(data) == 0:
            print(f"❌ Masque vide ignoré : {filename}")
            nb_vides += 1
            continue

        # Copier le fichier s'il est utile
        output_path = os.path.join(output_dir, filename)
        shutil.copy2(filepath, output_path)
        print(f"✅ Masque conservé : {filename}")
        nb_conservés += 1

print(f"\n🎯 Résultat : {nb_conservés} masques conservés, {nb_vides} vides ignorés.")
```

```python
import os
import shutil

# === Dossiers à modifier ===
img_dir = "/home/amenacer/Stage/data-4D/hadia/TollSome nifti/3D_slices"
mask_dir = "/home/amenacer/Stage/data-4D/hadia/nifti-mask/G1_non_vides"
output_dir = "/home/amenacer/Stage/data-4D/hadia/TollSome nifti/images_filtrees1"
os.makedirs(output_dir, exist_ok=True)

# === Extraire les identifiants uniques depuis les masques ===
mask_basenames = set()
for mask_file in os.listdir(mask_dir):
    if mask_file.endswith(".nii") or mask_file.endswith(".nii.gz"):
        parts = mask_file.split("_")
        if len(parts) >= 2:
            prefix = parts[0]  # ex: TollC3S0B%7
            z_info = [p for p in parts if p.startswith("z")]
            if z_info:
                z = z_info[0]
                key = f"{prefix}_z{z[-2:]}"
                mask_basenames.add(key)

# === Parcourir les images et copier celles correspondant aux masques ===
nb_copies = 0
for img_file in os.listdir(img_dir):
    if img_file.endswith(".nii") or img_file.endswith(".nii.gz"):
        parts = img_file.split("_")
        if len(parts) >= 2:
            prefix = parts[0]
            z_info = [p for p in parts if p.startswith("z")]
            if z_info:
                z = z_info[0]
                key = f"{prefix}_{z}"
                if key in mask_basenames:
                    src = os.path.join(img_dir, img_file)
                    dst = os.path.join(output_dir, img_file)
                    shutil.copy2(src, dst)
                    print(f"✅ Copié : {img_file}")
                    nb_copies += 1

print(f"\n🎉 Terminé : {nb_copies} images ont été filtrées et copiées.")
```

```python
import os
import nibabel as nib
import numpy as np

# === Dossiers à adapter ===
input_dir = "/home/amenacer/Stage/data-4D/hadia/nifti-mask/G3_non_vides"
output_dir = "/home/amenacer/Stage/data-4D/hadia/nifti-mask/G3_fusion"
os.makedirs(output_dir, exist_ok=True)

nb_fusion = 0
nb_ignore = 0

for fname in os.listdir(input_dir):
    if "_ROIMask-1_" in fname:
        roi1_path = os.path.join(input_dir, fname)
        roi2_path = roi1_path.replace("_ROIMask-1_", "_ROIMask-2_")

        if not os.path.exists(roi2_path):
            print(f"⚠️ ROIMask-2 manquant pour : {fname}")
            nb_ignore += 1
            continue

        # Chargement des deux masques
        roi1_img = nib.load(roi1_path)
        roi2_img = nib.load(roi2_path)
        roi1_data = roi1_img.get_fdata()
        roi2_data = roi2_img.get_fdata()

        # On binarise pour éviter les valeurs égales dans les 2 masques
        roi1_bin = (roi1_data > 0).astype(np.uint8)
        roi2_bin = (roi2_data > 0).astype(np.uint8)

        # Création d'une image fusionnée
        fusion = np.zeros_like(roi1_bin)

        # Ventricule gauche = 1
        fusion[roi1_bin == 1] = 1

        # Myocarde = 2 (si non déjà pris par VG)
        fusion[(roi2_bin == 1) & (fusion == 0)] = 2

        # Sauvegarde
        fusion_img = nib.Nifti1Image(fusion, roi1_img.affine, roi1_img.header)
        output_name = fname.replace("_ROIMask-1_", "_FUSION_")
        nib.save(fusion_img, os.path.join(output_dir, output_name))
        print(f"✅ Fusion sauvegardée : {output_name}")
        nb_fusion += 1

print(f"\n🎯 {nb_fusion} fusions créées, {nb_ignore} ignorées.")
```

```python
import os
import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap

# === Dossiers ===
input_dir = "/home/amenacer/Stage/data-4D/hadia/nifti-mask/G3_fusion"
output_dir = "/home/amenacer/Stage/data-4D/hadia/nifti-mask/G3_visualisation"
os.makedirs(output_dir, exist_ok=True)

# === Colormap personnalisée : 0 = fond, 1 = ventricule, 2 = myocarde
cmap = ListedColormap(["black", "red", "blue"])

# === Parcourir tous les fichiers
for fname in os.listdir(input_dir):
    if fname.endswith(".nii.gz") and "FUSION" in fname:
        filepath = os.path.join(input_dir, fname)
        img = nib.load(filepath)
        data = img.get_fdata()

        # Choisir une slice centrale
        z = data.shape[2] // 2
        slice_img = data[:, :, z]

        # Création de l’image
        plt.figure(figsize=(5, 5))
        plt.imshow(slice_img, cmap=cmap, vmin=0, vmax=2)
        plt.title(fname)
        plt.axis("off")
        plt.tight_layout()

        # Sauvegarde
        output_img = os.path.join(output_dir, fname.replace(".nii.gz", ".png"))
        plt.savefig(output_img, dpi=150)
        plt.close()

        print(f"✅ Image enregistrée : {output_img}")

print("🎉 Terminé : toutes les images ont été générées.")
```
