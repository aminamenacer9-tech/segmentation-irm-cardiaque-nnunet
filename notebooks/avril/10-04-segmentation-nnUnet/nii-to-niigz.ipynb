```python
import os
import nibabel as nib

input_dir = "/home/amenacer/Stage/base_de_donnees/3D/3D-short"
output_dir = "/home/amenacer/Stage/base_de_donnees/3D/3D-short-gz"

os.makedirs(output_dir, exist_ok=True)

for fname in os.listdir(input_dir):
    if fname.endswith("_0000.nii"):
        full_path = os.path.join(input_dir, fname)
        img = nib.load(full_path)

        new_fname = fname + ".gz"  # devient .nii.gz
        nib.save(img, os.path.join(output_dir, new_fname))
        print(f"✅ Converti : {fname} → {new_fname}")
```
