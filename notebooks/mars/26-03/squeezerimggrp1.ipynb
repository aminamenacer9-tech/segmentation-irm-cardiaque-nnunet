```python
import nibabel as nib
import numpy as np

# === Chemin de l'image à corriger ===
input_path = '/home/amenacer/Stage/Data/tray/grp1/TollC4S3B_Original.nii.gz'

# === Charger l'image
img_nii = nib.load(input_path)
img_data = img_nii.get_fdata()

print("Avant squeeze :", img_data.shape)

# === Supprimer la dernière dimension si elle est égale à 1
img_data = np.squeeze(img_data, axis=2)

print("Après squeeze :", img_data.shape)

# === Sauvegarde
output_path = '/home/amenacer/Stage/Data/TollC4S3B_Original.nii.gz'
new_img = nib.Nifti1Image(img_data, affine=img_nii.affine, header=img_nii.header)
nib.save(new_img, output_path)

print("✅ Image sauvegardée en 3D :", output_path)
```
