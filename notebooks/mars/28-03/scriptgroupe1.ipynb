```python
import os
import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt
```

```python
images_dir = '/home/amenacer/Stage/Data/TollSome Segmentation /Segmentation/Tollsomemask/Groupe2'
masks_dir = '/home/amenacer/Stage/Data/TollSome Segmentation /Segmentation/Tollsomemask/Groupe2'
output_dir = '/home/amenacer/Stage/Data/TollSome Segmentation /Segmentation/resultas/tollsome 2D+T/maskedimg/Tollsome_output2'

os.makedirs(output_dir, exist_ok=True)
```

```python
# === Lister les fichiers d'images et traiter ===
for filename in os.listdir(images_dir):
    if filename.endswith('.nii.gz'):
        image_path = os.path.join(images_dir, filename)
        mask_path = os.path.join(masks_dir, filename.replace('_0000', ''))  
        output_path = os.path.join(output_dir, filename)

        print(f"Image : {image_path}")
        print(f"Masque : {mask_path}")
        print(f"Résultat : {output_path}")
        break  
```

```python
# Chargement image et masque 
img_nii = nib.load(image_path)
mask_nii = nib.load(mask_path)

img_data = img_nii.get_fdata()
mask_data = mask_nii.get_fdata()

# Application : produit pixel à pixel 
applied = img_data * mask_data
```

```python
# Sauvegarde du résultat
applied_nii = nib.Nifti1Image(applied, img_nii.affine, img_nii.header)
nib.save(applied_nii, output_path)

print(f"Fichier sauvegardé : {output_path}")
```

```python
#  Lister les fichiers d'images et traiter toutes
for filename in os.listdir(images_dir):
    if filename.endswith('.nii.gz'):
        image_path = os.path.join(images_dir, filename)
        mask_path = os.path.join(masks_dir, filename.replace('_0000', '')) 
        output_path = os.path.join(output_dir, filename)

        if not os.path.exists(mask_path):
            print(f"⚠️ Masque manquant pour : {filename}")
            continue
        # Chargement
        img_nii = nib.load(image_path)
        mask_nii = nib.load(mask_path)

        img_data = img_nii.get_fdata()
        mask_data = mask_nii.get_fdata()

        # Application
        applied = img_data * mask_data

        # Sauvegarde
        applied_nii = nib.Nifti1Image(applied, img_nii.affine, img_nii.header)
        nib.save(applied_nii, output_path)

        print(f"✅ Traitement terminé pour : {filename}")
```

```python
import matplotlib.pyplot as plt

# Nombre d'images à afficher
n = 2
count = 0

for filename in os.listdir(images_dir):
    if filename.endswith('.nii.gz') and count < n:
        image_path = os.path.join(images_dir, filename)
        mask_path = os.path.join(masks_dir, filename.replace('_0000', ''))
        output_path = os.path.join(output_dir, filename)

        if not os.path.exists(mask_path):
            continue

        # Chargement
        img = nib.load(image_path).get_fdata()
        mask = nib.load(mask_path).get_fdata()
        result = img * mask

        # On choisit une coupe centrale
        slice_idx = 10 

        # Affichage
        fig, axes = plt.subplots(1, 3, figsize=(15, 5))
        axes[0].imshow(img[:, :, slice_idx], cmap='gray')
        axes[0].set_title('Image originale')

        axes[1].imshow(mask[:, :, slice_idx], cmap='Reds', alpha=0.6)
        axes[1].set_title('Masque')

        axes[2].imshow(result[:, :, slice_idx], cmap='gray')
        axes[2].set_title('Image × Masque')

        for ax in axes:
            ax.axis('off')
        plt.tight_layout()
        plt.show()

        count += 1
```
