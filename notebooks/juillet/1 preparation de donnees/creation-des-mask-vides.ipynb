```python
import os
import nibabel as nib
import numpy as np

# === À adapter ===
input_dir = "/home/amenacer/0178/"      # Dossier contenant les 20 images .nii.gz
output_dir = "/home/amenacer/labelsTs/"     # Dossier où sauver les masques vides

os.makedirs(output_dir, exist_ok=True)

for file in os.listdir(input_dir):
    if not file.endswith(".nii.gz"):
        continue

    img_path = os.path.join(input_dir, file)
    img = nib.load(img_path)
    shape = img.shape  # ex: (128, 128, 30)
    affine = img.affine
    header = img.header

    # Masque vide
    empty_mask = np.zeros(shape, dtype=np.uint8)

    # Nom de sortie
    out_name = file.replace("_0000", "")
    out_path = os.path.join(output_dir, out_name)

    nib.save(nib.Nifti1Image(empty_mask, affine, header), out_path)
    print(f"✅ {out_name} — masque vide créé | shape: {shape}")
```
