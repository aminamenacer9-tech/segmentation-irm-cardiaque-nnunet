```python
import nibabel as nib
import numpy as np
import os


```

```python

dossier_originaux = "/home/amenacer/Stage/data-4D/4D/w15/slices_squeezed/"
dossier_modifies = "/home/amenacer/Stage/Data/nnUnet/nnUNet_raw/Dataset007_MICE_HEART_2D/imagesTr/"

fichiers_originaux = [f for f in os.listdir(dossier_originaux) if f.endswith('.nii') or f.endswith('.nii.gz')]
fichiers_modifies = [f for f in os.listdir(dossier_modifies) if f.endswith('.nii') or f.endswith('.nii.gz')]

```

```python
# Charger tous les originaux en mémoire
data_originaux = {}
for f in fichiers_originaux:
    img = nib.load(os.path.join(dossier_originaux, f))
    data = img.get_fdata()
    data_originaux[f] = data

def images_identiques(data1, data2):
    return data1.shape == data2.shape and np.array_equal(data1, data2)
```

```python

# Pour chaque image modifiée, trouver le match parfait (distance = 0)
correspondances = []
for f_mod in fichiers_modifies:
    img_mod = nib.load(os.path.join(dossier_modifies, f_mod))
    data_mod = img_mod.get_fdata()
    match = None
    for f_ori, data_ori in data_originaux.items():
        if images_identiques(data_mod, data_ori):
            match = f_ori
            break
    if match:
        correspondances.append((f_mod, match))
        print(f"{f_mod} ---> {match}")

```

```python
import csv
with open("correspondances_parfaites.csv", "w") as f:
    writer = csv.writer(f)
    writer.writerow(["image_modifiee", "image_originale"])
    writer.writerows(correspondances)
```
