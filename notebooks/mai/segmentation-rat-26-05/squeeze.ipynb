```python
import os
import nibabel as nib
import numpy as np

input_folder = "/home/amenacer/Stage/base_de_donnees/rats/img-orginal/"  
output_folder = "/home/amenacer/Stage/base_de_donnees/rats/img-squeezed/"     

if not os.path.exists(output_folder):
    os.makedirs(output_folder)

for i in range(1, 30):
    fname = f"rat{i}.nii"
    in_path = os.path.join(input_folder, fname)
    out_fname = f"rat{i}_0000.nii.gz"
    out_path = os.path.join(output_folder, out_fname)

    # Load image
    img = nib.load(in_path)
    data = img.get_fdata()

    # Squeeze la 3ème dimension si elle est nulle (axe=2)
    # Si ta shape est (128, 128, 0, t), après squeeze => (128, 128, t)
    data = np.squeeze(data)

    # Remets la data dans un Nifti
    new_img = nib.Nifti1Image(data, affine=img.affine, header=img.header)
    nib.save(new_img, out_path)

    print(f"{in_path} => {out_path} | shape finale: {data.shape}")

print("Traitement terminé.")
```
